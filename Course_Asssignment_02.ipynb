 Assignment 2: Comparison Operators

Create a Student Result Checker.

Store a student's name, marks and passing marks, then use comparison operators to check:

1. Are the marks equal to 50?
2. Are the marks greater than 50?
3. Are the marks less than 50?
4. Are the marks greater than or equal to 50?
5. Are the marks not equal to 50?

Make sure you understand what result each comparison produces.

```python
student_name = "Laiba Ajmal"
obtained_marks = 97
passing_marks = 50
print("Student Name: ", student_name)
print("Are the marks equal to 50?", obtained_marks == passing_marks)
print("Are the marks greater than 50?", obtained_marks > passing_marks)
print("Are the marks less than 50?", obtained_marks <= passing_marks)
print("Are the marks greater than or equal to 50?", obtained_marks >= passing_marks)
print("Are the marks not equal to 50?", obtained_marks != passing_marks)
```
